```python
# Section 1: Input Data

# January input data 
Jan_subscribers = 12500
Sub_fee = 2500
Jan_movies_watched = 85000
Jan_operating_exp = 18000000

# February input data 
Feb_subscribers = 15800
Feb_movies_watched = 110600
Feb_operating_exp = 21000000

# Section 2: Calculations

# Subcription Revenue
Jan_revenue = Jan_subscribers * Sub_fee
Feb_revenue = Feb_subscribers * Sub_fee
Total_revenue = Jan_revenue + Feb_revenue

# Net Profit
Jan_net_profit = Jan_revenue - Jan_operating_exp 
Feb_net_profit = Feb_revenue - Feb_operating_exp

# Percentage Growth

#Subscribers
Perc_Sub_Growth_Jan_Feb = ((Feb_subscribers - Jan_subscribers)/Jan_subscribers) * 100

# Revenue 
Perc_Rev_Growth_Jan_Feb = ((Feb_revenue - Jan_revenue)/Jan_revenue) * 100

# Average Movies Watched
Avg_movies_watched_per_sub_Jan = Jan_movies_watched/Jan_subscribers
Avg_movies_watched_per_sub_Feb = Feb_movies_watched/Feb_subscribers

# % Change in Net Profit
Perc_Change_Net_Profit = (Feb_net_profit - Jan_net_profit)/Jan_net_profit * 100

# Net Profit Combined
Combined_Net_Profit = Jan_net_profit + Feb_net_profit 

# Section 3: Display Results

print(f"Subscription Revenue for January: N{Jan_revenue:,}")
print(f"Subscription Revenue for Febrary: N{Feb_revenue:,}")
print(f"Net Profit for January: N{Jan_net_profit:,}")
print(f"Net Profit for February: N{Feb_net_profit:,}")
print(f"Percentage Growth in Subscribers: {Perc_Sub_Growth_Jan_Feb:.2f}%")
print(f"Percentage Growth in Revenue: {Perc_Rev_Growth_Jan_Feb:.2f}%")
print(f"Average Number of Movies Watched per Subscriber in January: {Avg_movies_watched_per_sub_Jan}")
print(f"Average Number of Movies Watched per Subscriber in February: {Avg_movies_watched_per_sub_Feb}")
print(f"Percentage Change in Net Profit: {Perc_Change_Net_Profit:.2f}%")
print(f"Combined Net Profit: N{Combined_Net_Profit:,}")
```
