```python
# Mini-project 3 — Pharmaceutical Distribution Performance Analysis

products = ["Paracetamol", "Amoxicillin", "Ibuprofen", "Vitamin C", "Cetirizine", "Metformin"]

unit_prices = [500, 1500, 800, 350, 650, 1200]

predicted_sales = [1200, 850, 950, 1500, 700, 600]

actual_sales = [1350, 780, -999, 1650, 620, 720]

available_stock = [1800, 1200, 1400, 2000, 1000, 950]

carton_capacity = 24

# The verified replacement for -999 is 1050.

# Section A - Data cleaning and validation

corrected_sales = actual_sales.copy()
corrected_sales[2]= 1050

# checking for truthiness
a = all(products)
b = all(unit_prices)
c = all(predicted_sales)
d = all(corrected_sales)
e = all(available_stock)

# validation results list
validation_list = list([a, b, c, d, e])

# Determining if all give passes
x = all(validation_list)

# Section B - Data organisation

# combining all data set lists
combined_lists = list(zip(products, unit_prices, predicted_sales, corrected_sales, available_stock))
# numbering the combined lists
numbered_combined = list(enumerate(combined_lists, start=1))
# product total
total_num_products = len(products)

# Section C - Sales performance analysis

# total predicted sales
total_pred_sales = sum(predicted_sales)

# total corrected sales
total_corr_sales = sum(corrected_sales)

# average sales per product 
avg_sales_pp = total_corr_sales / total_num_products

# best selling product
most_sold_vol = max(corrected_sales)
most_sold_ind = corrected_sales.index(most_sold_vol)
most_sold_product = products[most_sold_ind]
most_sold_product_price = unit_prices[most_sold_ind]

# least selling product
least_sold_vol = min(corrected_sales)
least_sold_ind = corrected_sales.index(least_sold_vol)
least_sold_product = products[least_sold_ind]

# best selling product's contribution
best_selling_perc_contribution = (most_sold_vol / total_corr_sales) * 100


# Section D - Forecasting analysis

# Calculating absolute difference between predicted and actual sales
abs_difference = abs(total_pred_sales - total_corr_sales)
# absolute percentage erroe
abs_perc_error = (abs_difference / total_corr_sales) * 100
# pairing predicted and actual corrected sales 
paired_sales = list(zip(predicted_sales, corrected_sales))

# Section E - Inventory and revenue analysis

# total available stock before sales
stock_b4_sales = sum(available_stock)

# total stock after sales 
stock_aft_sales = stock_b4_sales - total_corr_sales

# carton packaging
carton_filled = divmod(stock_aft_sales, carton_capacity)

# revenue generated by best selling product
rev_by_best_selling = most_sold_vol * most_sold_product_price

# three highest actual sales figure
corrected_sales_sorted = sorted(corrected_sales, reverse=True)

three_highest = corrected_sales_sorted[:3]

# Displaying results 
# Section A
print(f"Original sales list: {actual_sales}")
print(f"Corrected sales list: {corrected_sales}")
print(f"Validation result: {validation_list}")
print(f"All five datasets passed truthiness check: {x}")

# Section B
print(f"COmbined Inventory: {combined_lists}")
print(f"COmbined inventory with numbers: {numbered_combined}")
print(f"Total number of products: {total_num_products}")

# Section C
print(f"Total predicted sales: {total_pred_sales}")
print(f"Total corrected sales: {total_corr_sales}")
print(f"Average actual sales per product: {avg_sales_pp:.2f}")
print(f"Best selling product is {most_sold_product} with a sales volume of {most_sold_vol} sold at {most_sold_product_price} per unit")
print(f"Least selling product is {least_sold_product} with a sales volume of {least_sold_vol} units")
print(f"Best selling product: {most_sold_product} contributed {best_selling_perc_contribution:.2f}% to the total sales")

# Section D
print(f"The absolute difference between predicted and actual sales: {abs_difference}")
print(f"The absolute percentage error: {abs_perc_error:.2f}%")
print(f"Paired list: {paired_sales}")

# Section E
print(f"Total available stock before sales: {stock_b4_sales} units")
print(f"Total available stock after sales: {stock_aft_sales} units")
print(f"Packaging results: {carton_filled[0]} cartons filled, {carton_filled[1]} units unpacked")
print(f"Total revenue by best selling product: ₦{rev_by_best_selling:,.2f}")
print(f"Three highest sales in descending order: {three_highest}")
```
