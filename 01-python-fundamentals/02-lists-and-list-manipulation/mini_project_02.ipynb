```python
# Movie Catalogue Performance Analysis

# Input data



titles = [
    "Inception", "Interstellar", 
    "Titanic", "Avatar", 
    "Gladiator", "The Dark Knight", 
    "Joker", "The Matrix"
]

release_years = [
    2010, 2014, 1997, 2009,
    2000, 2008, 2019, 1999
]

durations = [
   148, 169, 195, 162,
   155, 152, 122, 136
]

ratings = [
    8.8, 8.7, 7.9, 7.8, 
    8.5, 9.0, 8.4, 8.7
]

views = [
    45000, 52000, 38000, 61000, 
    29000, 58000, 41000, 35000
]


# A: Catalogue Overview

# Count the total number of movies in the titles list
total_num_movies = len(titles)

# divide the sum of the movie durations by total number of movies to get the average duration
duration_sum = sum(durations)
avg_movie_duration = duration_sum/total_num_movies

# B: Movie duration analysis

#Identify the longest movie (find longest duration, find its index, find corresponding title)
longest_duration = max(durations)
longest_movie_ind = durations.index(longest_duration)
longest_movie_title = titles[longest_movie_ind]

# Identify shortest movie (find shortest duration, find its index, find corresponding title)
shortest_duration = min(durations)
shortest_movie_ind = durations.index(shortest_duration)
shortest_movie_title = titles[shortest_movie_ind]

# C: Ratings and popularity

# Identify highest rated movie.
highest_rating = max(ratings)
highest_rating_ind = ratings.index(highest_rating)
high_rated_movie = titles[highest_rating_ind]
highest_rated_movie_release_year = release_years[highest_rating_ind]

# Most watched movie
most_watched = max(views)
most_watched_ind = views.index(most_watched)
most_watched_movie = titles[most_watched_ind]
most_watched_movie_rating = ratings[most_watched_ind]

# D: Viewership analysis 

# Sum of all views
total_movie_views = sum(views)
avg_movie_views = total_movie_views / total_num_movies

# Percentage contribution of most watched to total views
most_watched_perc_contribution = (most_watched/total_movie_views) * 100

# E: Statistical analysis 

# retrieve the three highest movie ratings without modifying original ratings list
# it would require a duplicate of the original list
sorted_ratings = sorted(ratings) 
three_highest_ratings = sorted_ratings[-3:]

# Difference between highest and lowest will require max and min values in ratings list
highest_to_lowest_rating_diff = max(ratings) - min(ratings)

# Displaying Results
print(f"Total number of movies: {total_num_movies}")
print(f"Average movie duration: {avg_movie_duration:.1f}")
print(f"The longest movie is: {longest_movie_title}. It runs for {longest_duration} minutes")
print(f"The shortest movie is: {shortest_movie_title}. It runs for {shortest_duration} minutes")
print(f"The Highest rated movie is: {high_rated_movie}. It has a {highest_rating} rating")
print(f"The most watched movie is: {most_watched_movie}. It currently has {most_watched_movie_rating} rating")
print(f"The total number of views currently sits at: {total_movie_views:,}. Average views per movie: {avg_movie_views:,}")
print(f"The most watched movie, {most_watched_movie}, contributes: {most_watched_perc_contribution:.2f}% to the total views")
print(f"The three highest ratings: {three_highest_ratings}")
print(f"The highest and lowest rated movies have a ratings difference of {highest_to_lowest_rating_diff:.2f}")
```
