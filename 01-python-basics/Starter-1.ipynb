```python
product_price = "4500.75"
units_sold = "25"
discount = "0.10"

price_aft_discount = float(product_price) - (float(product_price) * float(discount))

total_revenue = (price_aft_discount * int(units_sold))
                 
print("The total revenue is: " + str(round(total_revenue)))
```

```python
tablets = 850
sold = 327
discarded = 18

remaining_stock = tablets - (sold + discarded)

perc_sold = round((sold/tablets) * 100, 2)
perc_disc = round((discarded/tablets) * 100, 2)

print("The remaining stock is:", remaining_stock, "tablets")
print("Percentage of stock sold is:", perc_sold, "%")
print("Percentage of stock discarded is:", perc_disc, "%")
```

```python
cost_price = 2400
selling_price = 3200
qty = 150
transport = 25000
advert = 15000

total_exp = (cost_price * qty) + transport + advert
total_revenue = selling_price * qty

net_profit = total_revenue - total_exp
net_profit_margin = round((net_profit/total_revenue) * 100, 2)

print("The Total Expenditure = " + str(total_exp))
print("The Total Revenue = " + str(total_revenue))
print("The Net Profit = " + str(net_profit))
print("The Net Profit Margin = " + str(net_profit_margin) + "%")
```

```python
pop_2020 = 125000
pop_2025 = 158000

abs_incr = pop_2025 - pop_2020
perc_incr = (abs_incr/pop_2020) * 100

cAGR = ((((pop_2025/pop_2020) ** 0.2) - 1) * 100)

print("The Absolute Increase is: " + str(abs_incr))
print("The Percentage Increase is: " + str(perc_incr) + "%")
print("The Compound Annual Growth Rate is: " + str(round(cAGR, 1)) + "%")
```
